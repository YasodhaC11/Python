```python
#9.What will happen when you run the following code?


count = 10

def increment():
    count += 1
    print(count)

increment()

# count variable is not defined. variable defined outside the function cannot be accessed inside a function. Else we have to declare that as a global variable

count = 10

def increment():
    global count
    count += 1
    print(count)

increment()

```
